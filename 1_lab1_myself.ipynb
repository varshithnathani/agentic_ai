```python
from openai import OpenAI

client = OpenAI(
    base_url = "http://localhost:11434/v1",
    api_key = "ollama"
)

response = client.chat.completions.create(
    model = "llama3.2",
    messages= [{"role":"user", "content":"what is 2+2?"}]
)

print(response)
print(response.choices[0].message.content)
```

Get a question from the model and ask answer for that question 

```python
message = "get a difficult question to check the IQ of the model. Just give the question, don't give the answer."
response = client.chat.completions.create(
    model = "llama3.2",
    messages = [{"role":"user", "content":message}]
)

# print(response.choices[0].message.content)
question = response.choices[0].message.content
print(question)
```
